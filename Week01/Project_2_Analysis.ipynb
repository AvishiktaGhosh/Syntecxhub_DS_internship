```python
import pandas as pd
df = pd.read_csv("athlete_events.csv")
print(df.head())
print(df.info())
print(df.describe())
print(df.tail())
```

```python
df.drop_duplicates(inplace=True)
ndf=df.dropna()
```

```python
ndf['Age']=ndf["Age"].astype(int)
ndf['ID']=ndf["ID"].astype(int)
print(ndf.info())
```

```python
ndf=ndf[ndf["Age"]<=40] #filtering
print(ndf.info())
```

```python
print(ndf.loc[0:100,"Games"]) #slicing
```

```python
ndf.to_csv("athlete_events_final.csv", index=False)
```
